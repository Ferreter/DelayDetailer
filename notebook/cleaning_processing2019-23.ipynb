```python
#Import necessary libraries and data paths

import pandas as pd
import numpy as np
from pathlib import Path

pd.set_option("display.max_columns", None)

RAW_PATH = Path("../dataset/raw/flights_sample_3m.csv")
PROCESSED_DIR = Path("../dataset/processed")
PROCESSED_DIR.mkdir(parents=True, exist_ok=True)
```

```python

#load the raw data

df = pd.read_csv(RAW_PATH, low_memory=False)

print("Raw shape:", df.shape)
display(df.head())
display(df.info())
```

```python
# Clean column names: strip whitespace, convert to uppercase, replace spaces with underscores
df.columns = (
    df.columns
    .str.strip()
    .str.upper()
    .str.replace(" ", "_")
)

print(df.columns)
```

```python
# Select only the columns we need for analysis and visualization
cols_to_keep = [
    "FL_DATE",
    "AIRLINE",
    "AIRLINE_CODE",
    "FL_NUMBER",

    "ORIGIN",
    "ORIGIN_CITY",
    "DEST",
    "DEST_CITY",

    "CRS_DEP_TIME",
    "DEP_TIME",
    "DEP_DELAY",
    "TAXI_OUT",
    "WHEELS_OFF",

    "WHEELS_ON",
    "TAXI_IN",
    "CRS_ARR_TIME",
    "ARR_TIME",
    "ARR_DELAY",

    "CANCELLED",
    "CANCELLATION_CODE",
    "DIVERTED",

    "CRS_ELAPSED_TIME",
    "ELAPSED_TIME",
    "AIR_TIME",
    "DISTANCE",

    "DELAY_DUE_CARRIER",
    "DELAY_DUE_WEATHER",
    "DELAY_DUE_NAS",
    "DELAY_DUE_SECURITY",
    "DELAY_DUE_LATE_AIRCRAFT"
]

existing_cols = [col for col in cols_to_keep if col in df.columns]
missing_cols = [col for col in cols_to_keep if col not in df.columns]

df = df[existing_cols].copy()

print("Kept columns:", len(existing_cols))
print("Missing columns:", missing_cols)
print("Shape after selection:", df.shape)
display(df.head())
```

```python
print("Rows before date cleaning:", df.shape[0])
print("Sample dates:")
display(df["FL_DATE"].head())

df["FL_DATE"] = pd.to_datetime(df["FL_DATE"], errors="coerce") #convert to datetime, invalid parsing will be set as NaT
 
print("Invalid dates:", df["FL_DATE"].isna().sum())

df = df.dropna(subset=["FL_DATE"]).copy() #drop rows where date is invalid (NaT)

df["YEAR"] = df["FL_DATE"].dt.year
df["MONTH"] = df["FL_DATE"].dt.month
df["MONTH_NAME"] = df["FL_DATE"].dt.month_name()
df["DAY"] = df["FL_DATE"].dt.day
df["DAY_OF_WEEK"] = df["FL_DATE"].dt.dayofweek + 1
df["DAY_NAME"] = df["FL_DATE"].dt.day_name()
df["QUARTER"] = df["FL_DATE"].dt.quarter

print("Rows after date cleaning:", df.shape[0])
print("Date range:", df["FL_DATE"].min(), "to", df["FL_DATE"].max())
display(df["YEAR"].value_counts().sort_index())
```

```python
# Convert numeric columns to appropriate types, coercing errors to NaN
numeric_cols = [
    "FL_NUMBER",
    "CRS_DEP_TIME",
    "DEP_TIME",
    "DEP_DELAY",
    "TAXI_OUT",
    "WHEELS_OFF",
    "WHEELS_ON",
    "TAXI_IN",
    "CRS_ARR_TIME",
    "ARR_TIME",
    "ARR_DELAY",
    "CANCELLED",
    "DIVERTED",
    "CRS_ELAPSED_TIME",
    "ELAPSED_TIME",
    "AIR_TIME",
    "DISTANCE",
    "DELAY_DUE_CARRIER",
    "DELAY_DUE_WEATHER",
    "DELAY_DUE_NAS",
    "DELAY_DUE_SECURITY",
    "DELAY_DUE_LATE_AIRCRAFT",
    "YEAR",
    "MONTH",
    "DAY",
    "DAY_OF_WEEK",
    "QUARTER"
]

for col in numeric_cols:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors="coerce")

categorical_cols = [
    "AIRLINE",
    "AIRLINE_CODE",
    "ORIGIN",
    "ORIGIN_CITY",
    "DEST",
    "DEST_CITY",
    "CANCELLATION_CODE",
    "MONTH_NAME",
    "DAY_NAME"
]
# Convert categorical columns to string type
for col in categorical_cols:
    if col in df.columns:
        df[col] = df[col].astype("string")
```

```python

delay_cause_cols = [
    "DELAY_DUE_CARRIER",
    "DELAY_DUE_WEATHER",
    "DELAY_DUE_NAS",
    "DELAY_DUE_SECURITY",
    "DELAY_DUE_LATE_AIRCRAFT"
]

# Delay causes are blank when no delay reason was recorded.
df[delay_cause_cols] = df[delay_cause_cols].fillna(0)

# status columns: fill missing with 0 (not cancelled/diverted) and convert to int
df["CANCELLED"] = df["CANCELLED"].fillna(0).astype(int)
df["DIVERTED"] = df["DIVERTED"].fillna(0).astype(int)

# Cancellation code exists only for cancelled flights.
df["CANCELLATION_CODE"] = df["CANCELLATION_CODE"].fillna("Not cancelled")
```

```python

print("Missing values before operational cleaning:")
display(df.isna().sum().sort_values(ascending=False).head(20))

# Keep cancelled flights for cancellation analysis.
cancelled_df = df[df["CANCELLED"] == 1].copy()
non_cancelled_df = df[df["CANCELLED"] == 0].copy()

required_operational_cols = [
    "DEP_DELAY",
    "ARR_DELAY",
    "TAXI_OUT",
    "TAXI_IN",
    "ELAPSED_TIME",
    "AIR_TIME"
]

non_cancelled_df = non_cancelled_df.dropna(subset=required_operational_cols)

df = pd.concat([non_cancelled_df, cancelled_df], ignore_index=True)

# For cancelled flights, operational values are structurally missing.
operational_fill_zero = [
    "DEP_TIME",
    "DEP_DELAY",
    "TAXI_OUT",
    "WHEELS_OFF",
    "WHEELS_ON",
    "TAXI_IN",
    "ARR_TIME",
    "ARR_DELAY",
    "ELAPSED_TIME",
    "AIR_TIME"
]

df[operational_fill_zero] = df[operational_fill_zero].fillna(0)

print("Shape after operational cleaning:", df.shape)

print("Missing values after operational cleaning:")
display(df.isna().sum().sort_values(ascending=False).head(20))
```

```python
# Create new features for analysis and visualization

df["ROUTE"] = df["ORIGIN"] + " → " + df["DEST"]

df["DEP_DELAY_MINUTES"] = df["DEP_DELAY"].clip(lower=0) #negative delays (early departures) are set to 0 for delay analysis
df["ARR_DELAY_MINUTES"] = df["ARR_DELAY"].clip(lower=0) #negative delays (early arrivals) are set to 0 for delay analysis

df["IS_DELAYED"] = (df["ARR_DELAY"] >= 15).astype(int) #flights with arrival delay of 15 minutes or more are considered delayed
df["IS_DEP_DELAYED"] = (df["DEP_DELAY"] >= 15).astype(int) #flights with departure delay of 15 minutes or more are considered departure delayed
df["IS_CANCELLED"] = (df["CANCELLED"] == 1).astype(int) #flights with CANCELLED=1 are marked as cancelled
df["IS_DIVERTED"] = (df["DIVERTED"] == 1).astype(int)# flights with DIVERTED=1 are marked as diverted
def season_from_month(month):
    if month in [12, 1, 2]:
        return "Winter"
    elif month in [3, 4, 5]:
        return "Spring"
    elif month in [6, 7, 8]:
        return "Summer"
    else:
        return "Autumn"

df["SEASON"] = df["MONTH"].apply(season_from_month)
# Define delay severity based on arrival delay minutes
def delay_severity(minutes):
    if minutes <= 0:
        return "On time / Early"
    elif minutes < 15:
        return "Minor delay"
    elif minutes < 60:
        return "Moderate delay"
    else:
        return "Severe delay"

df["DELAY_SEVERITY"] = df["ARR_DELAY_MINUTES"].apply(delay_severity)
# Create a total delay cause column by summing all individual causes
df["TOTAL_DELAY_CAUSE"] = df[delay_cause_cols].sum(axis=1)

def dominant_delay_cause(row):
    causes = {
        "Carrier": row["DELAY_DUE_CARRIER"],
        "Weather": row["DELAY_DUE_WEATHER"],
        "NAS": row["DELAY_DUE_NAS"],
        "Security": row["DELAY_DUE_SECURITY"],
        "Late Aircraft": row["DELAY_DUE_LATE_AIRCRAFT"]
    }

    if sum(causes.values()) == 0:
        return "No recorded cause"

    return max(causes, key=causes.get)

df["DOMINANT_DELAY_CAUSE"] = df.apply(dominant_delay_cause, axis=1)

cancel_map = {
    "A": "Carrier",
    "B": "Weather",
    "C": "National Air System",
    "D": "Security",
    "Not cancelled": "Not cancelled"
}
# Map cancellation codes to reasons, with a default for non-cancelled flights
df["CANCELLATION_REASON"] = df["CANCELLATION_CODE"].map(cancel_map).fillna("Not cancelled")

def punctuality_status(row):
    if row["CANCELLED"] == 1:
        return "Cancelled"
    elif row["DIVERTED"] == 1:
        return "Diverted"
    elif row["ARR_DELAY"] >= 15:
        return "Delayed"
    else:
        return "On time"

df["PUNCTUALITY_STATUS"] = df.apply(punctuality_status, axis=1)

display(df.head())
```

```python
# Convert categorical columns to category dtype for memory efficiency
category_cols = [
    "AIRLINE",
    "AIRLINE_CODE",
    "ORIGIN",
    "ORIGIN_CITY",
    "DEST",
    "DEST_CITY",
    "CANCELLATION_CODE",
    "MONTH_NAME",
    "DAY_NAME",
    "ROUTE",
    "SEASON",
    "DELAY_SEVERITY",
    "DOMINANT_DELAY_CAUSE",
    "CANCELLATION_REASON",
    "PUNCTUALITY_STATUS"
]

for col in category_cols:
    if col in df.columns:
        df[col] = df[col].astype("category")

print(df.info(memory_usage="deep"))
```

```python
# Create summary tables for Shiny app
airline_summary = (
    df.groupby(["AIRLINE_CODE", "AIRLINE"], observed=True)
    .agg(
        total_flights=("FL_NUMBER", "count"),
        avg_dep_delay=("DEP_DELAY_MINUTES", "mean"),
        avg_arr_delay=("ARR_DELAY_MINUTES", "mean"),
        delay_rate=("IS_DELAYED", "mean"),
        cancellation_rate=("IS_CANCELLED", "mean"),
        diversion_rate=("IS_DIVERTED", "mean"),
        avg_distance=("DISTANCE", "mean"),
        avg_air_time=("AIR_TIME", "mean")
    )
    .reset_index()
)
# Summarize by origin airport and city
airport_summary = (
    df.groupby(["ORIGIN", "ORIGIN_CITY"], observed=True)
    .agg(
        total_departures=("FL_NUMBER", "count"),
        avg_dep_delay=("DEP_DELAY_MINUTES", "mean"),
        avg_arr_delay=("ARR_DELAY_MINUTES", "mean"),
        delay_rate=("IS_DELAYED", "mean"),
        cancellation_rate=("IS_CANCELLED", "mean"),
        avg_taxi_out=("TAXI_OUT", "mean")
    )
    .reset_index()
)
# Summarize by destination airport and city
route_summary = (
    df.groupby(["ORIGIN", "DEST", "ROUTE"], observed=True)
    .agg(
        total_flights=("FL_NUMBER", "count"),
        avg_dep_delay=("DEP_DELAY_MINUTES", "mean"),
        avg_arr_delay=("ARR_DELAY_MINUTES", "mean"),
        delay_rate=("IS_DELAYED", "mean"),
        cancellation_rate=("IS_CANCELLED", "mean"),
        avg_distance=("DISTANCE", "mean"),
        avg_air_time=("AIR_TIME", "mean")
    )
    .reset_index()
)

monthly_summary = (
    df.groupby(["YEAR", "MONTH", "MONTH_NAME"], observed=True)
    .agg(
        total_flights=("FL_NUMBER", "count"),
        avg_dep_delay=("DEP_DELAY_MINUTES", "mean"),
        avg_arr_delay=("ARR_DELAY_MINUTES", "mean"),
        delay_rate=("IS_DELAYED", "mean"),
        cancellation_rate=("IS_CANCELLED", "mean"),
        diversion_rate=("IS_DIVERTED", "mean")
    )
    .reset_index()
)

yearly_summary = (
    df.groupby("YEAR", observed=True)
    .agg(
        total_flights=("FL_NUMBER", "count"),
        avg_dep_delay=("DEP_DELAY_MINUTES", "mean"),
        avg_arr_delay=("ARR_DELAY_MINUTES", "mean"),
        delay_rate=("IS_DELAYED", "mean"),
        cancellation_rate=("IS_CANCELLED", "mean"),
        diversion_rate=("IS_DIVERTED", "mean")
    )
    .reset_index()
)

weekday_summary = (
    df.groupby(["DAY_OF_WEEK", "DAY_NAME"], observed=True)
    .agg(
        total_flights=("FL_NUMBER", "count"),
        avg_arr_delay=("ARR_DELAY_MINUTES", "mean"),
        delay_rate=("IS_DELAYED", "mean"),
        cancellation_rate=("IS_CANCELLED", "mean")
    )
    .reset_index()
)

delay_cause_summary = (
    df.groupby(["YEAR", "MONTH"], observed=True)[delay_cause_cols]
    .sum()
    .reset_index()
)

status_summary = (
    df.groupby("PUNCTUALITY_STATUS", observed=True)
    .agg(total_flights=("FL_NUMBER", "count"))
    .reset_index()
)
```

```python
# Basic quality checks on the cleaned dataset
print("Cleaned dataset shape:", df.shape)

print("\nDate range:")
print(df["FL_DATE"].min(), "to", df["FL_DATE"].max())

print("\nYear distribution:")
display(df["YEAR"].value_counts().sort_index())

print("\nMissing values:")
display(df.isna().sum().sort_values(ascending=False).head(20))

print("\nPunctuality status distribution:")
display(df["PUNCTUALITY_STATUS"].value_counts())

print("\nDelay severity distribution:")
display(df["DELAY_SEVERITY"].value_counts())

print("\nTop airlines:")
display(
    airline_summary
    .sort_values("total_flights", ascending=False)
    .head(10)
)

print("\nWorst routes with at least 500 flights:")
display(
    route_summary
    .query("total_flights >= 500")
    .sort_values("avg_arr_delay", ascending=False)
    .head(10)
)
```

```python
# Export cleaned dataset and summary tables to CSV for Shiny app
df.to_csv(PROCESSED_DIR / "flights_cleaned.csv", index=False)

airline_summary.to_csv(PROCESSED_DIR / "airline_summary.csv", index=False)
airport_summary.to_csv(PROCESSED_DIR / "airport_summary.csv", index=False)
route_summary.to_csv(PROCESSED_DIR / "route_summary.csv", index=False)
monthly_summary.to_csv(PROCESSED_DIR / "monthly_summary.csv", index=False)
yearly_summary.to_csv(PROCESSED_DIR / "yearly_summary.csv", index=False)
weekday_summary.to_csv(PROCESSED_DIR / "weekday_summary.csv", index=False)
delay_cause_summary.to_csv(PROCESSED_DIR / "delay_cause_summary.csv", index=False)
status_summary.to_csv(PROCESSED_DIR / "status_summary.csv", index=False)

print("CSV export complete.")
print("Files saved to:", PROCESSED_DIR)
```

```python
#airport map summary with airport lookup data for mapping in Shiny app
AIRPORTS_PATH = Path("../dataset/raw/airports.csv")

airports_lookup = pd.read_csv(AIRPORTS_PATH)

airports_lookup = airports_lookup[
    [
        "iata_code",
        "name",
        "latitude_deg",
        "longitude_deg",
        "iso_country",
        "type"
    ]
].copy()

airports_lookup = airports_lookup.dropna(subset=["iata_code", "latitude_deg", "longitude_deg"])

# Keep US airports only, since our flight dataset is US domestic
airports_lookup = airports_lookup[airports_lookup["iso_country"] == "US"]

airports_lookup = airports_lookup.rename(
    columns={
        "iata_code": "ORIGIN",
        "name": "AIRPORT_NAME",
        "latitude_deg": "LATITUDE",
        "longitude_deg": "LONGITUDE",
        "type": "AIRPORT_TYPE"
    }
)

airport_map_summary = airport_summary.merge(
    airports_lookup,
    on="ORIGIN",
    how="left"
)

airport_map_summary = airport_map_summary.dropna(subset=["LATITUDE", "LONGITUDE"])

print("Airport map summary shape:", airport_map_summary.shape)
display(airport_map_summary.head())
```
